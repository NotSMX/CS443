**Answer 2:** With r_seed=0, the RNN generates "shoute:" instead of "shouted:" — dropping the 'd'. This happens because the RNN outputs a softmax probability distribution over all vocabulary characters at each time step, and the next character is sampled from that distribution rather than always taking the highest-probability character. Even when 'd' has the highest probability, there is still a nonzero chance of drawing a different character. A different random seed changes which sample is drawn at that step, and once a wrong character is chosen, the GRU state evolves differently for all subsequent time steps, causing the deviation to persist and compound.

As generation length increases, the compounding effect of even one wrong sampled character shifts the GRU hidden state onto a trajectory the network hasn't seen during training. From that point, the state no longer corresponds cleanly to any memorized context, so the probability distributions become less peaked and sampling produces increasingly incoherent output.

```
                # to the param name
                for param_name in layer_params:
                    params[layer.get_name() + '/'+ param_name] = layer_params[param_name]

            layer = layer.get_prev_layer_or_block()
        np.savez_compressed(full_path, **params)
```
